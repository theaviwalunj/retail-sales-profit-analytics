```python
df_clean["Order Date"] = pd.to_datetime(df_clean["Order Date"])
df_clean["Ship Date"] = pd.to_datetime(df_clean["Ship Date"])

print(df_clean[["Order Date", "Ship Date"]].dtypes)
```

```python
df_clean["Order Year"] = df_clean["Order Date"].dt.year
```

```python
df_clean["Order Month"] = df_clean["Order Date"].dt.month
```

```python
df_clean["Month Name"] = df_clean["Order Date"].dt.month_name()
```

```python
df_clean["Quarter"] = df_clean["Order Date"].dt.quarter
```

```python
df_clean[["Order Date", "Order Year", "Order Month", "Month Name", "Quarter"]].head()
```

```python
df_clean["Profit Margin"] = (
    df_clean["Profit"] / df_clean["Sales"]
) * 100
```

```python
df_clean[["Sales", "Profit", "Profit Margin"]].head()
```

```python
df_clean.info()
```

```python
total_sales = df_clean["Sales"].sum()
total_profit = df_clean["Profit"].sum()
total_orders = df_clean["Order ID"].nunique()

print("Total Sales:", round(total_sales, 2))
print("Total Profit:", round(total_profit, 2))
print("Total Orders:", total_orders)
```

```python
overall_margin = (total_profit / total_sales) * 100

print("Overall Profit Margin:", round(overall_margin, 2), "%")
```

```python
category_analysis = df_clean.groupby("Category").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum")
).sort_values("Sales", ascending=False)

category_analysis
```

```python
category_analysis["Profit Margin %"] = (
    category_analysis["Profit"] / category_analysis["Sales"]
) * 100

category_analysis
```

```python
region_analysis = df_clean.groupby("Region").agg(
    Sales=("Sales", "sum"),
    Profit=("Profit", "sum")
).sort_values("Sales", ascending=False)

region_analysis
```

```python
region_analysis["Profit Margin %"] = (
    region_analysis["Profit"] / region_analysis["Sales"]
) * 100

region_analysis
```

```python
query = """
SELECT
    "Product Name",
    SUM(Sales) AS Total_Sales,
    SUM(Profit) AS Total_Profit,
    ROUND(SUM(Profit) * 100.0 / SUM(Sales), 2) AS Profit_Margin
FROM sales
GROUP BY "Product Name"
HAVING SUM(Sales) > (
    SELECT AVG(Product_Sales)
    FROM (
        SELECT SUM(Sales) AS Product_Sales
        FROM sales
        GROUP BY "Product Name"
    )
)
AND SUM(Profit) < 0
ORDER BY Total_Sales DESC;
"""

result = pd.read_sql_query(query, conn)

result.head(10)
```

```python
import pandas as pd

df = pd.read_excel(
    r"C:\Users\KUNAL\Plain Text\sample_-_superstore.xls"
)

df_clean = df.copy()

print("Dataset loaded successfully!")
```

```python
print("===== BUSINESS INSIGHTS =====")

print("\n1. Overall Performance")
print("Total Sales:", round(df_clean["Sales"].sum(), 2))
print("Total Profit:", round(df_clean["Profit"].sum(), 2))
print("Profit Margin:", round(
    df_clean["Profit"].sum() / df_clean["Sales"].sum() * 100, 2
), "%")

print("\n2. Sales by Category")
print(
    df_clean.groupby("Category")[["Sales", "Profit"]]
    .sum()
    .sort_values("Sales", ascending=False)
)

print("\n3. Sales by Region")
print(
    df_clean.groupby("Region")[["Sales", "Profit"]]
    .sum()
    .sort_values("Sales", ascending=False)
)

print("\n4. Top 10 Products by Sales")
print(
    df_clean.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(10)
)

print("\n5. Top 10 Loss-Making Products")
print(
    df_clean.groupby("Product Name")["Profit"]
    .sum()
    .sort_values()
    .head(10)
)
```

```python
# Business Insights Summary

category_summary = (
    df_clean.groupby("Category")[["Sales", "Profit"]]
    .sum()
    .sort_values("Sales", ascending=False)
)

region_summary = (
    df_clean.groupby("Region")[["Sales", "Profit"]]
    .sum()
    .sort_values("Sales", ascending=False)
)

top_product = (
    df_clean.groupby("Product Name")["Sales"]
    .sum()
    .sort_values(ascending=False)
    .head(1)
)

loss_product = (
    df_clean.groupby("Product Name")["Profit"]
    .sum()
    .sort_values()
    .head(1)
)

print("===== KEY BUSINESS INSIGHTS =====")

print("\n1. Best Sales Category:")
print(category_summary.index[0])

print("\n2. Best Sales Region:")
print(region_summary.index[0])

print("\n3. Top Product by Sales:")
print(top_product)

print("\n4. Most Loss-Making Product:")
print(loss_product)
```
